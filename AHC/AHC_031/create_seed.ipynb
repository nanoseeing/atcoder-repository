```python
from pathlib import Path 

file_name = Path("seeds.txt")
if not file_name.exists():
    file_name.touch()

with open(file_name, "w") as file:
    for i in range(10000):
        file.write(f"{i}\n")
```

```python
import numpy as np
IN_DIR = Path("tools") / "in"

sum_d = []
ave_d = []
for txt_file in IN_DIR.iterdir():
    with open(txt_file, mode="r") as file:
        lines = file.readlines()
    # read input
    W, D, N = map(int, lines[0].split())
    a = []
    
    now_sums = []
    print("---")
    for line, d in zip(lines[1:], range(D)):
        d = list(map(int, line.split()))
        a.append(d)
        sum_d.append(sum(d))
        now_sums.append(sum(d))
        print(sum(d))
    ave_d.append(np.average(now_sums))
```

```python
import matplotlib.pyplot as plt
plt.hist(sum_d, bins=100)
plt.show()
```

```python
import subprocess
from pathlib import Path

N = 100
EXE_PATH = Path("./rule_solver/target/release/rule_solver.exe")
for i in range(N):
    input_txt_path = str(Path(f"./in/{str(i).zfill(4)}.txt").absolute())
    output_txt_path = f"./out/{str(i).zfill(4)}.txt"
    p = subprocess.Popen(
        [str(EXE_PATH)],
        stdin=subprocess.PIPE,
        stdout=subprocess.PIPE
    )
    with open(input_txt_path, "r") as f:
        input_str = f.read()
    with open(output_txt_path, "w") as f:
        output_str = p.communicate(input_str.encode())[0].decode()
        f.write(output_str)
```

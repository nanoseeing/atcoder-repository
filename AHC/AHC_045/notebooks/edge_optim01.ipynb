```python
import numpy as np
from scipy.optimize import minimize
```

```python
import time

# パラメータ
M = 800
mu = np.array(np.random.rand(M))  # 各辺の事前平均
sigma = np.ones(M)  # 各辺の標準偏差

# 制約: l0 < l1, l3 < l4
constraints = []
for _ in range(30):
    constraints.append((np.random.randint(0, M), np.random.randint(0, M)))


# 目的関数（負の対数事前分布、つまりMAP推定の目的関数）
def objective(l):
    return np.sum((l - mu) ** 2 / (2 * sigma**2))


# 制約条件を scipy 形式に変換（不等式制約: l_j - l_i > 0）
scipy_constraints = [{"type": "ineq", "fun": lambda l, i=i, j=j: l[j] - l[i] - 1e-6} for i, j in constraints]

# 初期値（事前平均をそのまま使う）
init = mu.copy()

# 最適化
st = time.perf_counter()
result = minimize(objective, init, constraints=scipy_constraints)
et = time.perf_counter() - st
print("最適化にかかった時間:", et)

# 結果の表示
if result.success:
    print("推定された辺の長さ:", result.x)
else:
    print("最適化に失敗しました:", result.message)
```

```python
from common import UnionFind

uf = UnionFind([0, 1, 2, 3, 4])

print("初期状態:")
print(uf)
print("グループ数:", uf.group_count())
print()

print("0と1を連結")
uf.union(0, 1)
print(uf)
print("0と1は同じグループ？", uf.same(0, 1))
print("グループサイズ(0):", uf.size(0))
print("グループ数:", uf.group_count())
print()

print("1と2を連結")
uf.union(1, 2)
print(uf)
print("0と2は同じグループ？", uf.same(0, 2))
print("グループサイズ(2):", uf.size(2))
print("グループ数:", uf.group_count())
print()

print("3と4を連結")
uf.union(3, 4)
print(uf)
print("3と4は同じグループ？", uf.same(3, 4))
print("グループ数:", uf.group_count())
print()

print("全グループのメンバー:")
print(uf.all_group_members())
```

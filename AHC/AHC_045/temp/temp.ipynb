```python
import matplotlib.pyplot as plt
import numpy as np

# x軸の範囲（-2から2まで）
x = np.linspace(-50, 10, 400)

# 指数関数 y = exp(x)


# プロット
plt.plot(x, np.exp(x / 100), label="y = exp(x/100)")
plt.plot(x, np.exp(x) / 100, label="y = exp(x)/100")
plt.title("Exponential Function")
plt.xlabel("x")
plt.ylabel("exp(x)")
plt.ylim(0, 5)
plt.grid(True)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

# パラメータ
mu = 0  # 平均
sigma = 300  # 標準偏差

# xの範囲を設定
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)

# 正規分布の確率密度関数を計算
y = norm.pdf(x, mu, sigma)

# グラフを描画
plt.plot(x, y, label=f"N({mu}, {sigma}^2)")
plt.title("std")
plt.xlabel("x")
plt.ylabel("rfd")
plt.legend()
plt.grid(True)
plt.show()
```

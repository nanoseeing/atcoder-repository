```python
tmp = 31851  

diffs = []
for x in range(1, 1000):
    for y in range(x, 1000):
        area = x * y
        if tmp + 100 > area >=tmp:
            diffs.append((area - tmp, x, y))
diffs = sorted(sorted(diffs)[:10], key=lambda x: x[1])
```

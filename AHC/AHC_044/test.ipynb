```python
import math
import matplotlib.pyplot as plt

def cooling_schedule(initial_temp, min_temp, elapsed_time, max_time):
    """
    時間に基づく指数減衰の冷却スケジュール
    T(t) = T_min + (T_max - T_min) * exp(-λ * t)
    λ は max_time の範囲で温度を min_temp にするように調整
    """
    if max_time is None:
        return initial_temp  # 時間制限なしの場合、温度は変えない

    lambda_param = math.log(initial_temp / min_temp) / max_time
    return min_temp + (initial_temp - min_temp) * math.exp(-lambda_param * elapsed_time)
```

```python
import time
initial_temp = 1000
min_temp = 0.1
max_time = 3
now_time = time.perf_counter()
temps = []
elapsed_times = []
while time.perf_counter() - now_time < max_time:
    elapsed_time = time.perf_counter() - now_time
    elapsed_times.append(elapsed_time)
    temps.append(cooling_schedule(initial_temp, min_temp, elapsed_time, max_time))
    time.sleep(0.1)
```

```python
# グラフ描画
plt.plot(elapsed_times, temps, marker='o')
```

```python
def exponential_schedule(init, obj, elapsed_time, max_time):
    """
    時間に基いて指数関数的に上昇するスケジュール
    """
    lambda_param = math.log(obj / init) / max_time
    return init * math.exp(lambda_param * elapsed_time)
```

```python
import time
init = 100
obj = 1
max_time = 0.3
now_time = time.perf_counter()
temps = []
elapsed_times = []
while time.perf_counter() - now_time < max_time:
    elapsed_time = time.perf_counter() - now_time
    elapsed_times.append(elapsed_time)
    temps.append(exponential_schedule(init, obj, elapsed_time, max_time))
    time.sleep(0.01)
```

```python
# グラフ描画
plt.plot(elapsed_times, temps, marker='o')
```
